# ClickHouse Connection Check

Notebook simples para validar a conectividade com o cluster ClickHouse usando as variáveis definidas em `.env` (via `config.settings`).

```python
from connectors.clickhouse_client import ClickHouseClient
from config.settings import clickhouse_config

display_config = {
    "host": clickhouse_config.host,
    "port": clickhouse_config.port,
    "database": clickhouse_config.database,
    "secure": clickhouse_config.secure,
    "user": clickhouse_config.user,
}
display_config
```

```python
client = None
try:
    client = ClickHouseClient()
    print("Conexão estabelecida com sucesso.")

    query = "SELECT 1 AS ok, version() AS version, now() AS executed_at"
    result = client.execute_query_with_result(query)
    rows = result.result_rows
    if not rows:
        raise RuntimeError("A consulta não retornou linhas.")

    print("Resultado da consulta de verificação:")
    for row in rows:
        print(row)
except Exception as exc:
    print(f"Falha ao conectar ou consultar o ClickHouse: {exc}")
finally:
    if client:
        client.close()
```
